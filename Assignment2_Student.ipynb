## 2. Prepare the data (3 marks)
Write `preprocess(raw)`:
- Put positive rows first, then negative rows, for each feature set.
- Create labels: positive = 1, negative = 0.
- If a row has a NaN or infinite value in any of the three feature sets, remove it from all three sets and the labels. Keep valid zeros.
- Print the number of removed rows.

Return `clean, y`. `clean` is a dictionary of the three unscaled feature arrays; `y` is the matching label array. There are **108 features** in total.

**Normalization is part of these 3 marks, but do it inside the training pipeline in Section 5.** Do not scale the full dataset here.


```python
def preprocess(raw):
    # 1. Combine positive and negative observations for each feature set
    clean = {}

    for feature in ["FS-I", "FS-II", "FS-IV"]:
        positive = raw["pos"][feature]
        negative = raw["neg"][feature]
        clean[feature] = np.concatenate((positive, negative), axis=0)

    # 2. Create labels: 1 for positive, 0 for negative
    n_pos = raw["pos"]["FS-I"].shape[0]
    n_neg = raw["neg"]["FS-I"].shape[0]

    y = np.concatenate((
        np.ones(n_pos, dtype=int),
        np.zeros(n_neg, dtype=int)
    ))

    # 3. Find observations with NaN or infinite values
    #    across ANY of the three feature sets
    valid = np.ones(len(y), dtype=bool)

    for feature in clean:
        valid &= np.isfinite(clean[feature]).all(axis=1)

    # 4. Count and remove invalid observations from all feature sets
    removed = np.sum(~valid)
    print(f"Number of removed rows: {removed}")

    for feature in clean:
        clean[feature] = clean[feature][valid]

    y = y[valid]

    # 5. Return unscaled features and labels
    return clean, y

print(preprocess(raw))
```
